# Expense Tracker App - Exploratory Analysis

This notebook demonstrates the end-to-end expense tracking pipeline:
- Synthetic data generation
- Data cleaning
- EDA and analysis
- Visualization
- Insights

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os

sns.set_style("whitegrid")

PROJECT_ROOT = os.path.dirname(os.path.abspath(''))
DATA_DIR = os.path.join(PROJECT_ROOT, "data")
print("Project root:", PROJECT_ROOT)
```

```python
# Load cleaned expense data
clean_path = os.path.join(DATA_DIR, "expenses_clean.csv")
if os.path.exists(clean_path):
    df = pd.read_csv(clean_path, parse_dates=["date"])
    print(f"Loaded {len(df)} expense records")
    print(df.head())
    print(df.dtypes)
else:
    print("Cleaned data not found. Run: python main.py")
```

```python
# EDA Summary
print("Total Spend:", df["amount"].sum())
print("Average Transaction:", df["amount"].mean())
print("Date Range:", df["date"].min(), "to", df["date"].max())
print("\nSpend by Category:")
print(df.groupby("category")["amount"].sum().sort_values(ascending=False))
```

```python
# Category spend bar chart
category_spend = df.groupby("category")["amount"].sum().sort_values(ascending=False)
plt.figure(figsize=(10, 6))
sns.barplot(x=category_spend.values, y=category_spend.index, palette="Set2")
plt.title("Total Spend by Category")
plt.xlabel("Total Spend (INR)")
plt.ylabel("Category")
plt.tight_layout()
plt.show()
```
